```python
class MonkeyBanana:
    def __init__(self):
        self.state = {
            "monkey": "Window",
            "box": "Window",
            "monkey_on": "Floor",
            "banana": False
        }

    def walk(self):
        print("Monkey walks to the box")
        self.state["monkey"] = "Window"

    def push_box(self):
        print("Monkey pushes the box to the center")
        self.state["box"] = "Center"
        self.state["monkey"] = "Center"

    def climb_box(self):
        print("Monkey climbs onto the box")
        self.state["monkey_on"] = "Box"

    def grasp_banana(self):
        if (self.state["monkey"] == "Center" and
            self.state["box"] == "Center" and
            self.state["monkey_on"] == "Box"):
            print("Monkey grasps the banana")
            self.state["banana"] = True
        else:
            print("Monkey cannot grasp the banana")

    def solve(self):
        print("Initial State:", self.state)
        self.walk()
        self.push_box()
        self.climb_box()
        self.grasp_banana()
        print("Final State:", self.state)

        if self.state["banana"]:
            print("Goal achieved: Monkey has the banana")
        else:
            print("Goal not achieved")

problem = MonkeyBanana()
problem.solve()
```

```python
N = 4
board =[[0 for _ in range(N)] for _ in range(N)]
def is_safe(row,col):
  for i in range(row):
    if board[i][col]==1:
      return False
  i=row -1
  j=col -1
  while i>=0 and j>=0:
    if board[i][j]==1:
      return False
    i-=1
    j-=1
  i=row -1
  j=col +1
  while i>=0 and j<N:
    if board[i][j]==1:
      return False
    i-=1
    j+=1
  return True
def solve(row):
  if row==N:
    return True
  for col in range(N):
    if is_safe(row,col):
      board[row][col]=1
      if solve(row+1):
        return True
      board[row][col]=0
  return False
if solve(0):
  print("Solution for 4-Queens Problem:\n")
  for row in board:
    print(" ".join("Q" if cell ==1 else "." for cell in row))
else:
  print("No solution exists")
```
